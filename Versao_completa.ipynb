```python
# Detecção por Z-Score

def detect_outliers_zscore(df, features, threshold=3):
    """Detecta outliers usando Z-score (desvio padrão)"""
    from scipy.stats import zscore

    z_scores = np.abs(stats.zscore(df[features]))
    outliers_mask = (z_scores > threshold).any(axis=1)

    outlier_features = []
    for i, feat in enumerate(features):
        mask = z_scores[:, i] > threshold
        outlier_features.append({
            'Feature': feat,
            'Outliers': mask.sum(),
            'Percentual': f"{mask.sum()/len(df)*100:.2f}%"
        })

    return outliers_mask, pd.DataFrame(outlier_features)

# Aplicando Z-Score
outliers_zscore, df_outliers_zscore = detect_outliers_zscore(df, features)

print("\n" + "=" * 60)
print("OUTLIERS DETECTADOS - Método Z-Score (threshold=3)")
print("=" * 60)
print(df_outliers_zscore.to_string(index=False))
print(f"\nTotal de registros com outliers (Z-score): {outliers_zscore.sum()} ({outliers_zscore.sum()/len(df)*100:.2f}%)")
```

```python
### Detecção por IQR (Interquartile Range)

def detect_outliers_iqr(df, features):
    """Detecta outliers usando o método IQR"""
    outliers_mask = pd.Series([False] * len(df))
    outlier_features = []

    for feat in features:
        Q1, Q3 = df[feat].quantile([0.25, 0.75])
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        mask = (df[feat] < lower_bound) | (df[feat] > upper_bound)
        outliers_mask = outliers_mask | mask
        outlier_features.append({
            'Feature': feat,
            'Outliers': mask.sum(),
            'Percentual': f"{mask.sum()/len(df)*100:.2f}%"
        })

    return outliers_mask, pd.DataFrame(outlier_features)

# Aplicando IQR
outliers_iqr, df_outliers_iqr = detect_outliers_iqr(df, features)

print("=" * 60)
print("OUTLIERS DETECTADOS - Método IQR (Interquartile Range)")
print("=" * 60)
print(df_outliers_iqr.to_string(index=False))
print(f"\nTotal de registros com outliers (IQR): {outliers_iqr.sum()} ({outliers_iqr.sum()/len(df)*100:.2f}%)")
```

```python
from sklearn.ensemble import IsolationForest

# Aplicando Isolation Forest
iso_forest = IsolationForest(contamination=0.1, random_state=42)
outliers_if = iso_forest.fit_predict(df[features]) == -1

print("\n" + "=" * 60)
print("OUTLIERS DETECTADOS - Método Isolation Forest")
print("=" * 60)
print(f"Total de registros detectados como outliers: {outliers_if.sum()} ({outliers_if.sum()/len(df)*100:.2f}%)")

# Comparando métodos
comparison = pd.DataFrame({
    'Método': ['IQR', 'Z-Score', 'Isolation Forest'],
    'Outliers': [outliers_iqr.sum(), outliers_zscore.sum(), outliers_if.sum()],
    'Percentual': [
        f"{outliers_iqr.sum()/len(df)*100:.2f}%",
        f"{outliers_zscore.sum()/len(df)*100:.2f}%",
        f"{outliers_if.sum()/len(df)*100:.2f}%"
    ]
})

print("\n" + "=" * 60)
print("RESUMO COMPARATIVO DOS MÉTODOS")
print("=" * 60)
print(comparison.to_string(index=False))
```

### 3.8 Decisão sobre Outliers

Após comparar os três métodos, optamos por remover apenas os registros com **consenso entre os três métodos** (IQR + Z-Score + Isolation Forest), representando ~7% do dataset.

> Essa abordagem conservadora preserva a maior parte dos dados — inclusive vinhos atípicos que podem ser genuinamente premium — e remove apenas os registros com maior probabilidade de erro de medição.

```python
from sklearn.ensemble import IsolationForest
from scipy import stats as scipy_stats

# Recalculando os três métodos para definir o consenso
# IQR
outliers_iqr_mask = pd.Series([False] * len(df))
for feat in features:
    Q1, Q3 = df[feat].quantile([0.25, 0.75])
    IQR = Q3 - Q1
    outliers_iqr_mask |= (df[feat] < Q1 - 1.5 * IQR) | (df[feat] > Q3 + 1.5 * IQR)

# Z-Score
z = np.abs(scipy_stats.zscore(df[features]))
outliers_zscore_mask = pd.Series((z > 3).any(axis=1), index=df.index)

# Isolation Forest
iso = IsolationForest(contamination=0.1, random_state=42)
outliers_if_mask = pd.Series(iso.fit_predict(df[features]) == -1, index=df.index)

# Consenso: detectado pelos três
consenso = outliers_iqr_mask & outliers_zscore_mask & outliers_if_mask
print(f"Outliers por consenso (3/3 métodos): {consenso.sum()} ({consenso.sum()/len(df)*100:.1f}%)")

# Removendo do dataframe
df = df[~consenso].reset_index(drop=True)
print(f"Dataset após remoção: {df.shape[0]} amostras")
```

## 3.5 Balanceamento com SMOTE

O dataset apresenta um desbalanceamento crítico (~86% Baixa/Média × ~14% Alta Qualidade).  
Para corrigir isso, aplicamos o **SMOTE** (*Synthetic Minority Oversampling Technique*), que gera exemplos **sintéticos** da classe minoritária interpolando pontos existentes — em vez de simplesmente duplicá-los.

> ⚠️ O SMOTE é aplicado **apenas no conjunto de treino**, após o split e a padronização, para evitar vazamento de informação (*data leakage*) no conjunto de teste.

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=7)
x_train_smote, y_train_smote = smote.fit_resample(x_train_scaled, y_train)

print("Distribuição ANTES do SMOTE:")
print(y_train.value_counts().rename({0: 'Baixa/Média', 1: 'Alta Qualidade'}).to_string())

print("\nDistribuição APÓS o SMOTE:")
import pandas as pd
pd.Series(y_train_smote).value_counts().rename({0: 'Baixa/Média', 1: 'Alta Qualidade'}).sort_index().pipe(print)

print(f"\nTamanho do treino original:  {len(y_train)}")
print(f"Tamanho do treino com SMOTE: {len(y_train_smote)}")
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

counts_before = y_train.value_counts().sort_index()
counts_after  = pd.Series(y_train_smote).value_counts().sort_index()

labels = ['Baixa/Média', 'Alta Qualidade']

axes[0].bar(labels, counts_before.values, color=['#d62728', '#2ca02c'], edgecolor='white')
axes[0].set_title('Antes do SMOTE (Treino)', fontweight='bold')
axes[0].set_ylabel('Frequência')
for bar, val in zip(axes[0].patches, counts_before.values):
    axes[0].text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5, str(val), ha='center', fontsize=11)

axes[1].bar(labels, counts_after.values, color=['#d62728', '#2ca02c'], edgecolor='white')
axes[1].set_title('Após o SMOTE (Treino)', fontweight='bold')
axes[1].set_ylabel('Frequência')
for bar, val in zip(axes[1].patches, counts_after.values):
    axes[1].text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5, str(val), ha='center', fontsize=11)

plt.suptitle('Efeito do SMOTE no Balanceamento das Classes', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 5. KNN — K-Nearest Neighbors

### Desempenho Geral

Ao longo deste notebook, treinamos e avaliamos quatro modelos de classificação para prever a qualidade de vinhos tintos com base em suas características físico-químicas. A tabela comparativa e a curva ROC mostram que os modelos baseados em árvores — especialmente o **Random Forest** — apresentaram o melhor desempenho geral, com maior AUC-ROC e melhor equilíbrio entre precisão e recall.

A **Regressão Logística**, apesar de ser o modelo mais simples, apresentou resultados competitivos e cumpriu bem seu papel de baseline, confirmando que as variáveis físico-químicas têm relação linear suficiente com a qualidade para sustentar uma boa separação das classes. O **KNN** teve desempenho intermediário — sensível à escolha de K e ao desbalanceamento das classes, mas robusto o suficiente para o problema.

### Pré-processamento e Tratamento de Outliers

Após a análise exploratória, aplicamos um tratamento conservador de outliers baseado no **consenso entre três métodos** (IQR, Z-Score e Isolation Forest). Apenas os registros identificados simultaneamente pelos três métodos (~7% do dataset) foram removidos, preservando vinhos atípicos que podem representar amostras genuinamente premium e eliminando apenas os casos com maior probabilidade de erro de medição.

### Feature Engineering

Três novas variáveis foram criadas a partir dos insights da EDA:

- **`alcohol_acid_ratio`** — razão entre álcool e acidez volátil, capturando a combinação do principal preditor positivo com o negativo
- **`free_so2_ratio`** — proporção do dióxido de enxofre livre em relação ao total, indicando eficiência de conservação
- **`high_alcohol`** — flag binária para teor alcoólico acima de 11%, faixa associada à maior qualidade nos boxplots

Essas features foram calculadas igualmente para os dados de treino, teste e nos casos reais de validação, garantindo consistência com o scaler ajustado.

### Ajuste do Random Forest

O Random Forest foi otimizado via **GridSearchCV** com validação cruzada de 5 folds, buscando maximizar o F1-score da classe Alta Qualidade. A busca explorou combinações de `n_estimators`, `max_depth` e `min_samples_leaf`, e o melhor estimador encontrado foi utilizado em todas as avaliações subsequentes.

### Variáveis Mais Relevantes

Os modelos convergiram na importância das mesmas variáveis identificadas na EDA, agora complementadas pelas features derivadas:

- **`alcohol`** e **`alcohol_acid_ratio`** — principais preditores positivos
- **`volatile acidity`** — principal preditor negativo, associada à deterioração do vinho
- **`sulphates`** — contribuição positiva consistente, ligada à conservação do produto
- **`free_so2_ratio`** — relevância moderada, associada à estabilidade microbiológica

### Balanceamento com SMOTE

O desbalanceamento das classes (~86% baixa/média × ~14% alta qualidade) foi tratado com **SMOTE** (*Synthetic Minority Oversampling Technique*), aplicado exclusivamente ao conjunto de treino após o split e a padronização — sem risco de data leakage. A comparação entre os modelos com e sem SMOTE evidenciou ganhos de Recall e F1 para a classe Alta Qualidade, que são as métricas mais relevantes para o problema. A acurácia global pode cair levemente com o SMOTE — comportamento esperado e saudável, pois o modelo original tinha acurácia inflada pela classe dominante.

### Teste com Casos Reais

O teste com registros reais do dataset confirmou a capacidade preditiva dos modelos em situações concretas. Os casos foram padronizados com o mesmo scaler do treino e com as features derivadas calculadas, garantindo que a predição replica exatamente o pipeline de produção.

### Implicações para o Processo Produtivo

Os resultados do modelo apontam para ações concretas que enólogos e produtores podem adotar para aumentar a proporção de vinhos de Alta Qualidade:

Teor alcoólico (alcohol) — a variável de maior impacto positivo. Vinhos com teor acima de 11% apresentaram proporção significativamente maior de Alta Qualidade. Na prática, isso sugere monitorar a fermentação para maximizar a conversão de açúcar, ajustando temperatura e tempo de fermentação conforme necessário.
Acidez volátil (volatile acidity) — principal preditor negativo, associada à produção de ácido acético por bactérias. Valores elevados indicam deterioração microbiológica. A intervenção recomendada é o controle rigoroso de higiene nas etapas de fermentação e armazenamento, além do monitoramento frequente durante o envelhecimento.
Sulfatos (sulphates) — contribuição positiva consistente. Atuam como agente conservante e antimicrobiano. O ajuste cuidadoso da dosagem de dióxido de enxofre (SO₂) durante o processo pode melhorar a estabilidade do produto final, respeitando os limites regulatórios.
Razão SO₂ livre / total (free_so2_ratio) — proporções mais altas indicam maior fração ativa do conservante disponível. Monitorar essa razão ao longo do envelhecimento permite intervir antes que a proteção antimicrobiana seja comprometida.

Em síntese, o modelo pode ser incorporado como um sistema de alerta precoce no processo produtivo: analisando as características físico-químicas em estágios intermediários da produção, o produtor pode identificar lotes com baixa probabilidade de Alta Qualidade e realizar ajustes antes do engarrafamento, reduzindo desperdício e aumentando a consistência do produto.

### Considerações Finais

Os resultados indicam que é possível construir um sistema de apoio à decisão confiável baseado em dados físico-químicos, reduzindo a dependência de avaliações sensoriais subjetivas e permitindo intervenções mais rápidas no processo de produção. Métricas como Recall e AUC-ROC devem ser priorizadas sobre a acurácia na avaliação final — um modelo que simplesmente classifique tudo como "baixa/média" já teria ~86% de acurácia sem nenhum poder preditivo real.
